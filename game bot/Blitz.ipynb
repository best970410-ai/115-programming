```python
import pyautogui as pyg

BOARD_ROWS = 8
BOARD_COLS = 8

def gem_center(row, col, basex, basey, step_x, step_y):
    """回傳第 row 列、第 col 欄寶石中心的螢幕座標（列、欄從 0 開始）。"""
    if not (0 <= row < BOARD_ROWS and 0 <= col < BOARD_COLS):
        raise ValueError("row 和 col 必須介於 0 到 7")
    if step_x <= 0 or step_y <= 0:
        raise ValueError("step_x 和 step_y 必須是正數像素間距")
    return basex + col * step_x, basey + row * step_y

def click_gem(row, col, basex, basey, step_x, step_y):
    """點擊 8×8 盤面中指定列、欄的寶石中心。"""
    x, y = gem_center(row, col, basex, basey, step_x, step_y)
    pyg.click(x, y)
```

```python
def gem_position(r, c):
    """回傳第 r 列、第 c 欄寶石中心的螢幕座標 (x, y)，列與欄從 0 開始。"""
    if not (0 <= r < BOARD_ROWS and 0 <= c < BOARD_COLS):
        raise ValueError("r 和 c 必須介於 0 到 7")
    x, y = gem_center(r, c, basex, basey, gap, gap)
    return round(x), round(y)
```

```python
# 自動三消：遊戲需先進入 8×8 棋盤。指定範圍為棋盤外框 (left, top, right, bottom)。
import random
import statistics
import time
import pyautogui as pyg

BOARD_ROWS = 8
BOARD_COLS = 8
BOARD_LEFT, BOARD_RIGHT = 495, 820
BOARD_TOP, BOARD_BOTTOM = 371, 691
GAME_DURATION_SECONDS = 60  # 若遊戲限時不同，請改成遊戲實際秒數。
COLOR_DISTANCE_THRESHOLD = 55
ANIMATION_WAIT_SECONDS = 0.45

pyg.FAILSAFE = True  # 執行中將滑鼠移至螢幕左上角可中止。

def _gem_screen_center(row, col):
    cell_width = (BOARD_RIGHT - BOARD_LEFT) / BOARD_COLS
    cell_height = (BOARD_BOTTOM - BOARD_TOP) / BOARD_ROWS
    x = BOARD_LEFT + (col + 0.5) * cell_width
    y = BOARD_TOP + (row + 0.5) * cell_height
    return round(x), round(y)

def _sample_rgb(image, x, y, radius=2):
    pixels = [image.getpixel((px, py))[:3]
              for px in range(x - radius, x + radius + 1)
              for py in range(y - radius, y + radius + 1)]
    return tuple(int(statistics.median(pixel[channel] for pixel in pixels))
                 for channel in range(3))

def _color_distance(a, b):
    return sum((a[channel] - b[channel]) ** 2 for channel in range(3)) ** 0.5

def read_board():
    image = pyg.screenshot()
    width, height = image.size
    if BOARD_RIGHT > width or BOARD_BOTTOM > height:
        raise ValueError(f"棋盤座標超出螢幕範圍：目前截圖為 {width}×{height}")

    samples = []
    for row in range(BOARD_ROWS):
        sample_row = []
        for col in range(BOARD_COLS):
            x, y = _gem_screen_center(row, col)
            rgb = _sample_rgb(image, x, y)
            # 排除棋盤底色等近黑、低彩度像素，避免將空格誤認成寶石。
            if max(rgb) < 42 and max(rgb) - min(rgb) < 20:
                rgb = None
            sample_row.append(rgb)
        samples.append(sample_row)

    clusters = []  # 每個元素為 [R總和, G總和, B總和, 樣本數]
    board = [[None] * BOARD_COLS for _ in range(BOARD_ROWS)]
    for row in range(BOARD_ROWS):
        for col in range(BOARD_COLS):
            rgb = samples[row][col]
            if rgb is None:
                continue
            nearest, nearest_distance = None, float("inf")
            for index, cluster in enumerate(clusters):
                center = tuple(value / cluster[3] for value in cluster[:3])
                distance = _color_distance(rgb, center)
                if distance < nearest_distance:
                    nearest, nearest_distance = index, distance
            if nearest is None or nearest_distance > COLOR_DISTANCE_THRESHOLD:
                clusters.append([rgb[0], rgb[1], rgb[2], 1])
                board[row][col] = len(clusters) - 1
            else:
                cluster = clusters[nearest]
                cluster[0] += rgb[0]
                cluster[1] += rgb[1]
                cluster[2] += rgb[2]
                cluster[3] += 1
                board[row][col] = nearest

    valid_cells = sum(cell is not None for row in board for cell in row)
    if valid_cells < 48 or not 3 <= len(clusters) <= 12:
        return None, len(clusters), valid_cells
    return board, len(clusters), valid_cells

def find_matches(board):
    matched = set()
    # 橫向連續相同顏色。
    for row in range(BOARD_ROWS):
        start = 0
        while start < BOARD_COLS:
            color = board[row][start]
            end = start + 1
            while color is not None and end < BOARD_COLS and board[row][end] == color:
                end += 1
            if color is not None and end - start >= 3:
                matched.update((row, col) for col in range(start, end))
            start = end
    # 縱向連續相同顏色。
    for col in range(BOARD_COLS):
        start = 0
        while start < BOARD_ROWS:
            color = board[start][col]
            end = start + 1
            while color is not None and end < BOARD_ROWS and board[end][col] == color:
                end += 1
            if color is not None and end - start >= 3:
                matched.update((row, col) for row in range(start, end))
            start = end
    return matched

def find_valid_swaps(board):
    moves = []
    existing_matches = find_matches(board)
    for row in range(BOARD_ROWS):
        for col in range(BOARD_COLS):
            for other_row, other_col in ((row, col + 1), (row + 1, col)):
                if other_row >= BOARD_ROWS or other_col >= BOARD_COLS:
                    continue
                if board[row][col] is None or board[other_row][other_col] is None:
                    continue
                if board[row][col] == board[other_row][other_col]:
                    continue
                board[row][col], board[other_row][other_col] = board[other_row][other_col], board[row][col]
                new_matches = find_matches(board) - existing_matches
                board[row][col], board[other_row][other_col] = board[other_row][other_col], board[row][col]
                if new_matches:
                    moves.append((len(new_matches), (row, col), (other_row, other_col)))
    if not moves:
        return []
    best_score = max(move[0] for move in moves)
    return [move for move in moves if move[0] == best_score]

started_at = time.monotonic()
invalid_frames = 0
no_move_signature = None
no_move_count = 0
move_count = 0
print("自動消除開始；將滑鼠移到螢幕左上角可緊急停止。")

try:
    while time.monotonic() - started_at < GAME_DURATION_SECONDS:
        board, color_count, valid_cells = read_board()
        if board is None:
            invalid_frames += 1
            if invalid_frames >= 5:
                print("棋盤無法辨識，可能遊戲已結束；停止操作。")
                break
            time.sleep(0.25)
            continue

        invalid_frames = 0
        signature = tuple(tuple(row) for row in board)
        best_moves = find_valid_swaps(board)
        if not best_moves:
            if signature == no_move_signature:
                no_move_count += 1
            else:
                no_move_signature, no_move_count = signature, 1
            if no_move_count >= 6:
                print("棋盤連續無可消除交換，停止操作。")
                break
            time.sleep(0.35)
            continue

        no_move_signature, no_move_count = None, 0
        _, (row1, col1), (row2, col2) = random.choice(best_moves)
        x1, y1 = _gem_screen_center(row1, col1)
        x2, y2 = _gem_screen_center(row2, col2)
        pyg.moveTo(x1, y1, duration=0.04)
        pyg.dragTo(x2, y2, duration=0.14, button="left")
        move_count += 1
        print(f"第 {move_count} 次交換：({row1}, {col1}) ↔ ({row2}, {col2})；辨識到 {color_count} 種顏色")
        time.sleep(ANIMATION_WAIT_SECONDS)
except pyg.FailSafeException:
    print("已觸發 PyAutoGUI 緊急停止。")

print(f"自動消除結束，共執行 {move_count} 次交換。")
```

```python
def _can_move_direction(r, c, dr, dc, im):
    """判斷交換到指定方向後，是否會形成新的三連或以上消除。"""
    board_size = 8
    next_r, next_c = r + dr, c + dc
    if not (0 <= r < board_size and 0 <= c < board_size
            and 0 <= next_r < board_size and 0 <= next_c < board_size):
        return False

    def color_at(row, col):
        x = round(basex + col * gap)
        y = round(basey + row * gap)
        return im.getpixel((x, y))[:3]

    def same_color(color_a, color_b):
        # 使用小幅 RGB 容差，降低截圖像素的細微差異造成的誤判。
        return sum((int(color_a[i]) - int(color_b[i])) ** 2 for i in range(3)) <= 35 ** 2

    def matched_positions(board):
        matches = set()
        for row in range(board_size):
            for col in range(board_size - 2):
                if (same_color(board[row][col], board[row][col + 1])
                        and same_color(board[row][col], board[row][col + 2])):
                    matches.update((row, col + offset) for offset in range(3))
        for col in range(board_size):
            for row in range(board_size - 2):
                if (same_color(board[row][col], board[row + 1][col])
                        and same_color(board[row][col], board[row + 2][col])):
                    matches.update((row + offset, col) for offset in range(3))
        return matches

    board = [[color_at(row, col) for col in range(board_size)]
             for row in range(board_size)]
    matches_before = matched_positions(board)
    board[r][c], board[next_r][next_c] = board[next_r][next_c], board[r][c]
    matches_after = matched_positions(board)
    return bool(matches_after - matches_before)

def can_move_left(r, c, im):
    return _can_move_direction(r, c, 0, -1, im)

def can_move_right(r, c, im):
    return _can_move_direction(r, c, 0, 1, im)

def can_move_up(r, c, im):
    return _can_move_direction(r, c, -1, 0, im)

def can_move_down(r, c, im):
    return _can_move_direction(r, c, 1, 0, im)

# 保留先前使用的函式名稱，等同於往左判斷。
can_move = can_move_left
```

```python
import time
import pyautogui as pyg

start_time = time.monotonic()
directions = (
    (can_move_right, 0, 1),
    (can_move_up, -1, 0),
    (can_move_down, 1, 0),
    (can_move_left, 0, -1),
)
move_count = 0

try:
    while True:
        if time.monotonic() - start_time >= 60:
            break

        im = pyg.screenshot()
        move = None
        for r in range(8):
            for c in range(8):
                for can_move_direction, dr, dc in directions:
                    if can_move_direction(r, c, im):
                        move = (r, c, r + dr, c + dc)
                        break
                if move is not None:
                    break
            if move is not None:
                break

        if move is None:
            time.sleep(0.1)
            continue

        r1, c1, r2, c2 = move
        x1, y1 = gem_position(r1, c1)
        x2, y2 = gem_position(r2, c2)
        pyg.moveTo(x1, y1, duration=0.04)
        pyg.dragTo(x2, y2, duration=0.14, button="left")
        move_count += 1
        time.sleep(0.3)  # 等待消除動畫完成，再重新截圖判斷。
except pyg.FailSafeException:
    print("已觸發 PyAutoGUI 緊急停止。")

print(f"一分鐘內共執行 {move_count} 次交換。")
```
