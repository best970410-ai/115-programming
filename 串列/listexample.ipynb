# 簡介
串列(list)是一個可以儲存多個元素(整數，浮點數，或字串)的容器。
串列中每個元素都是使用逗點隔開，左右有中括號[]

```python
scores = [78,29,56,80,99]
print(scores)
```

## 索引值
我們可以使用索引值(index)來存取串列中的元素。
注意:索引值由0開始

```python
print(scores[0])
print(scores[4])
```

## 串列內容的個數
len(a)取得a這個串列中元素的個數。
len 是(length)長度的意思

```python
print(len(scores))
print(scores[len(scores)-1])
print(scores[5-1])
```
