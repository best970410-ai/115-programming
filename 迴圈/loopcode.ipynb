# Loop
loop指令可以讓程式碼依我的條件不斷地執行，直到條件滿足為止。

```python
## 底下的程式碼會印出0-9
for i in range(10):
    print(i)
```

```python
list(range(10))
```

```python
for i in range(10):
    print(i)
```

# Example
計算 1+2+3....+10
                               

```python
sum =1
for i in range (10):
    sum = sum + (i+1)
print(sum)
```

# range
-懶惰是一種美德
-由0開始
-包含頭，不包含尾

```python
print(list(range(5)))
#區間，由0開始，到5-1結束
print(list(range(0,5)))
#區間，由1開始，到5-1結束
print(list(range(1,5)))
#區間，由1開始，到5-1結束，每次跳兩個
print(list(range(1,5,2)))
```

```python
print(list(range(1,100,2)))
```
